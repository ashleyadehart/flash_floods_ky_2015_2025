# Flash Floods KY Combined Data EDA

```python
# Import libraries and dependencies
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.transforms as mtransforms
import seaborn as sns
import pandas as pd
import numpy as np
import sqlite3
```

```python
# Read dataset into dataframe
df = pd.read_csv('../data/processed/flash_floods_ky_combined_data.csv')
```

```python
# Read first rows of dataset
df.head()
```

```python
# Read last rows of dataset
df.tail()
```

```python
# Check data type and missing values
df.info()
```

```python
# See the dimensions of the dataset
df.shape
```

```python
# Strip hidden spaces from all column names
df.columns = df.columns.str.strip()
```

```python
# Convert event_id to string
df['event_id'] = df['event_id'].astype(str)

df.info()
```

```python
# Convert date strings to datetime objects
df['begin_date'] = pd.to_datetime(df['begin_date'], format='mixed')
df['end_date'] = pd.to_datetime(df['end_date'], format='mixed')

df.info()
```

```python
# Standardize time ints into 4-digit strings
df['begin_time'] = df['begin_time'].astype(str).str.zfill(4)
df['end_time'] = df['end_time'].astype(str).str.zfill(4)

df.info()
```

```python
# Create unified datetime series
df['start_datetime'] = pd.to_datetime( df['begin_date'].dt.strftime('%Y-%m-%d') + ' ' + df['begin_time'], format='%Y-%m-%d %H%M' ) 
df['end_datetime'] = pd.to_datetime( df['end_date'].dt.strftime('%Y-%m-%d') + ' ' + df['end_time'], format='%Y-%m-%d %H%M' )
```

```python
# Move newly formed columns to after event_id column
start_datetime = df.pop('start_datetime')
end_datetime = df.pop('end_datetime')

df.insert(1, 'start_datetime', start_datetime)
df.insert(2, 'end_datetime', end_datetime)

df.info()
```

```python
# Drop columns
df = df.drop(columns=[
    'begin_date',
    'end_date',
    'begin_time',
    'end_time'
])

df.info()
```

```python
# Save dataframe to CSV file
df.to_csv('../data/processed/flash_floods_ky_combined_data.csv', index=False)
```
